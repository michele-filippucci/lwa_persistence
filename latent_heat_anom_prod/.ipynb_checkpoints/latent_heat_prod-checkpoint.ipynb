```python
ds = xr.open_dataset("siconc_clim_amip.nc",decode_times=False)
latent_heat_anom = ds["siconc_clim_amip"].mean('time')
latent_heat_anom.name = 'flux_lhe_anom'
ds = latent_heat_anom.to_dataset()
```

```python
#latent heat of evaporation (J/kg) 
#need to convert the value.
lhev = 2.5e6
```

```python
ds.assign_attrs(
    units="W/m^2", description="Latent Heat Flux"
)
```

```python
#PARAMETERS:
lat_ext = 50
intensity = 0 #units: W/m^2
```

```python
def gaussian(x, a, b, c):
    return a * np.exp(-(x - b)**2 / (2 * c**2))
```

```python
#initialize
latent_heat_anom.values = np.zeros(latent_heat_anom.shape)
#simple implementation: uniform polar flux
#latent_heat_anom.loc[90-lat_ext:,:] = intensity
#more complex: Gaussian heat
for i in range(lat_ext):
    latent_heat_anom.loc[90-0.5-i,:] = gaussian(90-i,intensity,90,lat_ext/2)
```

```python
lon = ds['lon']
lat = ds['lat']

plt.contourf(lon,lat,latent_heat_anom[:,:])
latent_heat_anom.shape
```

```python
ds.to_netcdf('flux_lhe_anom_'+str(intensity)+'_'+str(lat_ext)+'.nc')
```
