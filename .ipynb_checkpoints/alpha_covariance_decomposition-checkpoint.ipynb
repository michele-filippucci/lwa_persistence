```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np

# --- INPUT ---
# Assume you have xarray DataArrays: lwa and ucomp, with dims ('time','lat','lon')
# ref: reference period (e.g., control)
# pert: perturbed period (e.g., experiment)

pert = xr.open_dataset('/emmy-noether/home/mf671/DATA/isca_data/exeter_project_mf_aa_st_KS_30_72,90/tot/filtered_decomposition_data_flipped_2-51.nc')
ref = xr.open_dataset('/emmy-noether/home/mf671/DATA/isca_data/exeter_project_mf_baseline_st_KS/tot/filtered_decomposition_data_flipped_2-51.nc')

lat = pert['lat']
ucomp_ref = ref['ucomp']
ucomp_pert = pert['ucomp']
cosphi = np.cos(np.deg2rad(lat)).broadcast_like(ucomp_ref)
dw_lwa_ref = -cosphi*ref['dw_lwa']
dw_lwa_pert = -cosphi*pert['dw_lwa']
dw_lwa_var_ref = dw_lwa_ref.var(dim='time')
dw_lwa_var_pert = dw_lwa_pert.var(dim='time')

# --- 1. Means ---
E_X_ref = ucomp_ref.mean(dim='time') * np.sqrt(dw_lwa_var_ref)
E_Y_ref = dw_lwa_ref.mean(dim='time') * np.sqrt(dw_lwa_var_ref)
E_X_pert = ucomp_pert.mean(dim='time') * np.sqrt(dw_lwa_var_pert)
E_Y_pert = dw_lwa_pert.mean(dim='time') * np.sqrt(dw_lwa_var_pert)

# --- 2. Raw joint moments ---
E_XY_ref = (ucomp_ref*dw_lwa_ref).mean(dim='time') * dw_lwa_var_ref
E_XY_pert = (ucomp_pert*dw_lwa_pert).mean(dim='time') * dw_lwa_var_pert

# --- 3. Covariance decomposition ---
dCov_joint = E_XY_pert - E_XY_ref
dCov_Xmean = - E_Y_ref * (E_X_pert - E_X_ref)
dCov_Ymean = - E_X_ref * (E_Y_pert - E_Y_ref)
dCov_interaction = - (E_X_pert - E_X_ref) * (E_Y_pert - E_Y_ref)

# total decomposed change
dCov_total = dCov_joint + dCov_Xmean + dCov_Ymean + dCov_interaction

# --- 4. Actual covariance values ---
cov_ref = E_XY_ref - E_X_ref*E_Y_ref
cov_pert = E_XY_pert - E_X_pert*E_Y_pert
delta_cov = cov_pert - cov_ref
```

```python
#plot and sanity check

# --- 5. Sanity check ---
print("Max difference (should be ~0):", float((delta_cov - dCov_total).max()))

levels = np.arange(-600,600.2,40)

# --- 6. Plotting ---
terms = {
    'ΔCov joint': dCov_joint,
    'ΔCov ucomp mean': dCov_Xmean,
    'ΔCov dw_lwa mean': dCov_Ymean,
    'ΔCov interaction': dCov_interaction,
    'ΔCov total': dCov_total,
    'ΔCov actual': delta_cov
}

fig, axes = plt.subplots(2,3, figsize=(18,10), constrained_layout=True)
axes = axes.flatten()

for ax, (name, da) in zip(axes, terms.items()):
    im = da.plot(ax=ax, levels= levels, cmap='bwr', add_colorbar=True)
    ax.set_title(name)

plt.show()
```

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt

# --- Load data ---
pert = xr.open_dataset('/emmy-noether/home/mf671/DATA/isca_data/exeter_project_mf_aa_st_KS_30_72,90/tot/filtered_decomposition_data_flipped_2-51.nc')
ref  = xr.open_dataset('/emmy-noether/home/mf671/DATA/isca_data/exeter_project_mf_baseline_st_KS/tot/filtered_decomposition_data_flipped_2-51.nc')

# Shortcuts
lat = pert['lat']
u_ref = ref['ucomp']
u_pert = pert['ucomp']

# --- Apply cosine weighting to LWA tendency term ---
cosphi = np.cos(np.deg2rad(lat)).broadcast_like(u_ref)
lwa_ref = -cosphi * ref['dw_lwa']
lwa_pert = -cosphi * pert['dw_lwa']

# --- Compute variances and covariance ingredients (over time only!) ---
var_ref = lwa_ref.var(dim='time')
var_pert = lwa_pert.var(dim='time')

E_X_ref   = u_ref.mean(dim='time')
E_Y_ref   = lwa_ref.mean(dim='time')
E_X_pert  = u_pert.mean(dim='time')
E_Y_pert  = lwa_pert.mean(dim='time')

E_XY_ref  = (u_ref * lwa_ref).mean(dim='time')
E_XY_pert = (u_pert * lwa_pert).mean(dim='time')

# --- Covariance decomposition ---
dCov_joint       = E_XY_pert - E_XY_ref
dCov_Xmean       = -E_Y_ref * (E_X_pert - E_X_ref)
dCov_Ymean       = -E_X_ref * (E_Y_pert - E_Y_ref)
dCov_interaction = -(E_X_pert - E_X_ref) * (E_Y_pert - E_Y_ref)

dCov_total = dCov_joint + dCov_Xmean + dCov_Ymean + dCov_interaction

# --- Actual covariances ---
cov_ref  = E_XY_ref  - E_X_ref  * E_Y_ref
cov_pert = E_XY_pert - E_X_pert * E_Y_pert

# Sanity check
print("Sanity check (should be ~0):", float((cov_pert - cov_ref - dCov_total).max()))

# ======================================================
#       COMPUTE Δβ TERMS (beta = Cov / Var_LWA)
# ======================================================

beta_ref  = cov_ref  / var_ref
beta_pert = cov_pert / var_pert
dBeta     = beta_pert - beta_ref

# Apply to each decomposition term (Option A denominator = pert variance)
dBeta_joint       = dCov_joint       / var_pert
dBeta_Xmean       = dCov_Xmean       / var_pert
dBeta_Ymean       = dCov_Ymean       / var_pert
dBeta_interaction = dCov_interaction / var_pert
dBeta_total       = dCov_total       / var_pert

# ======================================================
#                   PLOT MAPS
# ======================================================

terms = {
    'Δβ (joint)': dBeta_joint,
    'Δβ (X mean shift)': dBeta_Xmean,
    'Δβ (Y mean shift)': dBeta_Ymean,
    'Δβ (interaction)': dBeta_interaction,
    'Δβ total (sum)': dBeta_total,
    'Δβ actual': dBeta,
}

fig, axes = plt.subplots(2,3, figsize=(18,10), constrained_layout=True)
axes = axes.flatten()

for ax, (title, da) in zip(axes, terms.items()):
    da.plot(ax=ax, cmap='bwr', center=0, add_colorbar=True)
    ax.set_title(title)

plt.show()
```
