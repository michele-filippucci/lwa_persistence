```python
#import dataset with fluxes (output of falwa)
data = xr.open_mfdataset('ERA5_northem_2.5x2.5_zg500_daily_190-1963.nc', parallel=True)

zg = data['zg']/9.81
lon = data['lon']
lat = data['lat']

data.close()
```

```python
zg
```

```python
data = mf.calculate_local_wave_activity(data, zg, monotonic_decreasing=True, planet_radius=6371.e3, n_quantity_levels=40)
```

```python
import cartopy.crs as ccrs
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10,5))
ax = plt.axes(projection=ccrs.PlateCarree())
cf = ax.contourf(lon, lat, zg[0], levels=np.arange(4500,6000,40), cmap='jet', transform=ccrs.PlateCarree())
plt.colorbar(cf, ax=ax, orientation='vertical', label='Geopotential Height (m)')
ax.coastlines()
ax.set_title('Geopotential Height (zg[0]) on PlateCarree Projection')
```

```python
import cartopy.crs as ccrs
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10,5))
ax = plt.axes(projection=ccrs.PlateCarree())
cf = ax.contourf(lon, lat, data.lwa[0], cmap='Reds', transform=ccrs.PlateCarree())
plt.colorbar(cf, ax=ax, orientation='vertical', label='LWA (m^2/s^2)')
ax.coastlines()
ax.set_title('LWA on PlateCarree Projection')
```
