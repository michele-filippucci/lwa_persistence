```python
import xarray as xr
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
import numpy as np
from tqdm import tqdm
import cftime
from dask.diagnostics import ProgressBar
```

```python
def flip_and_append(ds):
    # Step 2: Select Southern Hemisphere data
    ds_sh = ds.sel(lat=slice(-90,0))
    
    # Step 3: Flip the latitude values to Northern Hemisphere
    # Here, we need to change the latitude to positive values (0 to 90)
    ds_sh = ds_sh.assign_coords(lat = abs(ds_sh.lat))
    ds_sh = ds_sh.reindex(lat=list(reversed(ds_sh.lat)))
    
    date_list = []
    for date in ds_sh.time.values:
      date_list.append(cftime.Datetime360Day(date.year + ds_sh.time.values[-1].year, date.month, date.day))
    date_list = np.array(date_list)
    ds_sh = ds_sh.assign_coords(time = date_list)
    
    # Step 4: Concatenate the original dataset with the flipped Southern Hemisphere data
    # Assuming 'time' is the coordinate you want to concatenate along
    combined_ds = xr.concat([ds, ds_sh], dim='time')
    latitude = ds_sh['lat']
    ds_positive_latitudes = combined_ds.sel(lat=latitude[latitude >= 0])
    return ds_positive_latitudes
```

```python
ProgressBar().register()

# Path to data and experiments
base_path = '/home/links/mf671/DATA/isca_data'
experiments = [
    'exeter_project_mf_baseline_KS',
    'exeter_project_mf_aa_KS_30_72,90',
    'exeter_project_mf_baseline_st_KS',
    'exeter_project_mf_aa_st_KS_30_72,90'
]

# Load datasets
ds_clims = []
compute_climatology = False
if compute_climatology:
    for experiment in experiments:
        print(experiment)
        print(f'{base_path}/{experiment}/tot/TKE_monthly_2.5x2.5_2-51.nc')
        tke = flip_and_append(xr.open_mfdataset(f'{base_path}/{experiment}/tot/TKE_monthly_2.5x2.5_2-51.nc',
                                                parallel=True,
                                                chunks={'time': 30}))['tke'].loc[:,250,:,:].mean('time')
        tke_std = (flip_and_append(xr.open_mfdataset(f'{base_path}/{experiment}/tot/TKE_monthly_2.5x2.5_2-51.nc',
                                                parallel=True,
                                                chunks={'time': 30}))['tke'].loc[:,250,:,:].std('time')/np.sqrt(2*12*50)).rename('tke_std')
        dw_lwa = xr.open_mfdataset(f'{base_path}/{experiment}/tot/falwa_LWAA_tracked_flipped_2.5x2.5_daily_2-51.nc',
                                parallel=True,
                                chunks={'time': 30})['dw_lwa'].mean('time')
        dw_lwa_std = (xr.open_mfdataset(f'{base_path}/{experiment}/tot/falwa_LWAA_tracked_flipped_2.5x2.5_daily_2-51.nc',
                                parallel=True,
                                chunks={'time': 30})['dw_lwa'].std('time')/np.sqrt(2*365*50)).rename('dw_lwa_std')
        print(tke_std.shape,tke.shape,dw_lwa_std.shape,dw_lwa.shape)
        ds = xr.merge([tke,dw_lwa,tke_std,dw_lwa_std])
        ds.to_netcdf(f'{base_path}/{experiment}/TKE_dw_lwa_CLIM_1.5x1.5.nc')
        ds_clims.append(ds)
else:
    for experiment in experiments:
        ds = xr.open_dataset(f'{base_path}/{experiment}/TKE_dw_lwa_CLIM_1.5x1.5.nc')
        ds_clims.append(ds)

# Load blocking data
DA_clims = []

for experiment in tqdm(experiments):
    ds = xr.open_dataset(f'{base_path}/{experiment}/tot/falwa_LWAA_tracked_flipped_2.5x2.5_daily_2-51.nc')
    DA_clims.append((xr.where(ds['LWAA_tracked']>0,1,0).mean('time'))*100)
    ds.close()
```

```python
# python
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

# Initialize a 2-axis figure with Cartopy projection
fig, ax = plt.subplots(2, 2, figsize=(16, 6), subplot_kw={'projection': ccrs.PlateCarree()})
ax = ax.T.flatten()  # TRANSPOSE the 2x2 axes to invert rows and columns, then flatten for indexing

titles = ['symmetric','asymmetric']

# triangle coordinates
coords = [
    (-75.0, 25.0),  # Point A
    (-50.0, 25.0),  # Point B
    (-50.0, 50.0)   # Point C
]
# Unzip into separate lists of longitudes and latitudes, and close the triangle
lons, lats = zip(*coords)
lons += (lons[0],)
lats += (lats[0],)

for i in range(2):
    # Select axes and datasets based on iteration (using transposed/flatted axes)
    ax_tke = ax[i * 2]
    ax_lwa = ax[i * 2 + 1]
    ax_block = ax[i * 2 + 2]
    ds_a = ds_clims[i * 2]     # Experiment Baseline
    ds_b = ds_clims[i * 2 + 1] # Experiment Arctic Amplification
    da_a_block = DA_clims[i * 2]
    da_b_block = DA_clims[i * 2 + 1]

    # TKE plot
    levels = np.linspace(0, 100, 10)
    levels_diff = np.arange(-5.5, 6, 1)
    ax_tke.set_title(f'TKE (m^2/s^2) - {titles[i]} : AA (lime) vs Baseline (black)')
    c1 = ax_tke.contour(ds_a['lon'], ds_a['lat'], ds_a['tke'], levels=levels, colors='black', transform=ccrs.PlateCarree())
    c2 = ax_tke.contour(ds_b['lon'], ds_b['lat'], ds_b['tke'], levels=levels, colors='lime', transform=ccrs.PlateCarree())
    sh = ax_tke.contourf(ds_a['lon'], ds_a['lat'], ds_b['tke'] - ds_a['tke'],
                         levels=levels_diff, cmap='bwr', alpha=1, extend='both',
                         transform=ccrs.PlateCarree())
    fig.colorbar(sh, ax=ax_tke, orientation='vertical', label='TKE (m^2/s^2)')
    if i > 0:
        ax_tke.plot(lons, lats, linestyle='--', linewidth=3, color='red', transform=ccrs.PlateCarree())
    significance = np.abs(ds_b['tke'] - ds_a['tke']) > np.sqrt(ds_a['tke_std']**2 + ds_b['tke_std']**2)
    # optional significance hatch:
    # ax_tke.contourf(ds_a['lon'], ds_a['lat'], significance, levels=[0.5, 1], hatches=['...', None], colors='none', transform=ccrs.PlateCarree())
    gl0 = ax_tke.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.7, linestyle='--')
    gl0.top_labels = False
    gl0.right_labels = False

    # dw_lwa plot
    levels = np.linspace(5, 25, 8)
    levels_diff = np.arange(-2.25, 2.3, 0.3)
    ax_lwa.set_title(f'dw_lwa (W/m^2) - {titles[i]} : AA (lime) vs Baseline (black)')
    c3 = ax_lwa.contour(ds_a['lon'], ds_a['lat'], ds_a['dw_lwa'], levels=levels, colors='black', transform=ccrs.PlateCarree())
    c4 = ax_lwa.contour(ds_b['lon'], ds_b['lat'], ds_b['dw_lwa'], levels=levels, colors='lime', transform=ccrs.PlateCarree())
    sh2 = ax_lwa.contourf(ds_a['lon'], ds_a['lat'], ds_b['dw_lwa'] - ds_a['dw_lwa'],
                          levels=levels_diff, cmap='bwr', alpha=1, extend='both',
                          transform=ccrs.PlateCarree())
    fig.colorbar(sh2, ax=ax_lwa, orientation='vertical', label='dw_lwa (W/m^2)')
    if i > 0:
        ax_lwa.plot(lons, lats, linestyle='--', linewidth=3, color='red', transform=ccrs.PlateCarree())
    significance2 = np.abs(ds_b['dw_lwa'] - ds_a['dw_lwa']) > np.sqrt(ds_a['dw_lwa_std']**2 + ds_b['dw_lwa_std']**2)
    # optional significance hatch:
    # ax_lwa.contourf(ds_a['lon'], ds_a['lat'], significance2, levels=[0.5, 1], hatches=['...', None], colors='none', transform=ccrs.PlateCarree())
    gl1 = ax_lwa.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.7, linestyle='--')
    gl1.top_labels = False
    gl1.right_labels = False

    #blocking plot
    

plt.tight_layout()
plt.show()
```

```python
# python
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

# Figure with 3 rows and 2 columns, transpose to invert rows/columns, then flatten
fig, ax = plt.subplots(3, 2, figsize=(16, 12), subplot_kw={'projection': ccrs.PlateCarree()})
ax = ax.T.flatten()

titles = ['symmetric', 'asymmetric']

# triangle coordinates and closed polygon
coords = [(-75.0, 25.0), (-50.0, 25.0), (-50.0, 50.0)]
lons, lats = zip(*coords)
lons += (lons[0],)
lats += (lats[0],)

# First two rows: TKE and dw_lwa (same indexing pattern as provided)
for i in range(2):
    ax_tke = ax[i * 2]
    ax_lwa = ax[i * 2 + 1]
    ds_a = ds_clims[i * 2]
    ds_b = ds_clims[i * 2 + 1]

    # TKE plot
    levels_tke = np.linspace(0, 100, 10)
    levels_tke_diff = np.arange(-5.5, 6, 1)
    ax_tke.set_title(f'TKE (m^2/s^2) - {titles[i]} : AA (lime) vs Baseline (black)')
    ax_tke.contour(ds_a['lon'], ds_a['lat'], ds_a['tke'], levels=levels_tke, colors='black', transform=ccrs.PlateCarree())
    ax_tke.contour(ds_b['lon'], ds_b['lat'], ds_b['tke'], levels=levels_tke, colors='lime', transform=ccrs.PlateCarree())
    sh = ax_tke.contourf(ds_a['lon'], ds_a['lat'], ds_b['tke'] - ds_a['tke'],
                         levels=levels_tke_diff, cmap='bwr', alpha=1, extend='both',
                         transform=ccrs.PlateCarree())
    fig.colorbar(sh, ax=ax_tke, orientation='vertical', label='TKE (m^2/s^2)')
    if i > 0:
        ax_tke.plot(lons, lats, linestyle='--', linewidth=3, color='red', transform=ccrs.PlateCarree())
    significance_tke = np.abs(ds_b['tke'] - ds_a['tke']) > np.sqrt(ds_a['tke_std']**2 + ds_b['tke_std']**2)
    # Optional hatch for significance:
    # ax_tke.contourf(ds_a['lon'], ds_a['lat'], significance_tke, levels=[0.5, 1], hatches=['...', None], colors='none', transform=ccrs.PlateCarree())
    gl0 = ax_tke.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.7, linestyle='--')
    gl0.top_labels = False
    gl0.right_labels = False

    # dw_lwa plot
    levels_lwa = np.linspace(5, 25, 8)
    levels_lwa_diff = np.arange(-2.25, 2.3, 0.3)
    ax_lwa.set_title(f'dw_lwa (W/m^2) - {titles[i]} : AA (lime) vs Baseline (black)')
    ax_lwa.contour(ds_a['lon'], ds_a['lat'], ds_a['dw_lwa'], levels=levels_lwa, colors='black', transform=ccrs.PlateCarree())
    ax_lwa.contour(ds_b['lon'], ds_b['lat'], ds_b['dw_lwa'], levels=levels_lwa, colors='lime', transform=ccrs.PlateCarree())
    sh2 = ax_lwa.contourf(ds_a['lon'], ds_a['lat'], ds_b['dw_lwa'] - ds_a['dw_lwa'],
                          levels=levels_lwa_diff, cmap='bwr', alpha=1, extend='both',
                          transform=ccrs.PlateCarree())
    fig.colorbar(sh2, ax=ax_lwa, orientation='vertical', label='dw_lwa (W/m^2)')
    if i > 0:
        ax_lwa.plot(lons, lats, linestyle='--', linewidth=3, color='red', transform=ccrs.PlateCarree())
    significance_lwa = np.abs(ds_b['dw_lwa'] - ds_a['dw_lwa']) > np.sqrt(ds_a['dw_lwa_std']**2 + ds_b['dw_lwa_std']**2)
    # Optional hatch for significance:
    # ax_lwa.contourf(ds_a['lon'], ds_a['lat'], significance_lwa, levels=[0.5, 1], hatches=['...', None], colors='none', transform=ccrs.PlateCarree())
    gl1 = ax_lwa.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.7, linestyle='--')
    gl1.top_labels = False
    gl1.right_labels = False

# Third row: blocking frequency from DA_clims
for i in range(2):
    ax_block = ax[4 + i]  # indices 4 and 5 in flattened array
    da = DA_clims[i * 2]
    db = DA_clims[i * 2 + 1]

    # Blocking frequency levels (assumed 0..1)
    levels_block = np.linspace(0.0, 1.0, 6)
    levels_block_diff = np.linspace(-0.3, 0.3, 13)
    ax_block.set_title(f'Blocking frequency - {titles[i]} : AA (lime) vs Baseline (black)')
    ax_block.contour(da['lon'], da['lat'], da, levels=levels_block, colors='black', transform=ccrs.PlateCarree())
    ax_block.contour(db['lon'], db['lat'], db, levels=levels_block, colors='lime', transform=ccrs.PlateCarree())
    sh3 = ax_block.contourf(da['lon'], da['lat'], db - da,
                            levels=levels_block_diff, cmap='bwr', alpha=1, extend='both',
                            transform=ccrs.PlateCarree())
    fig.colorbar(sh3, ax=ax_block, orientation='vertical', label='Blocking frequency (fraction)')
    if i > 0:
        ax_block.plot(lons, lats, linestyle='--', linewidth=3, color='red', transform=ccrs.PlateCarree())
    #significance_block = np.abs(db - da) > np.sqrt(da['blocking_std']**2 + db['blocking_std']**2)
    # Optional hatch for significance:
    # ax_block.contourf(da['lon'], da['lat'], significance_block, levels=[0.5, 1], hatches=['...', None], colors='none', transform=ccrs.PlateCarree())
    glb = ax_block.gridlines(draw_labels=True, linewidth=0.5, color='gray', alpha=0.7, linestyle='--')
    glb.top_labels = False
    glb.right_labels = False

plt.tight_layout()
plt.show()
```
