```python
base_path = "/home/links/mf671/DATA/isca_data/"


experiments = [
    'exeter_project_mf_baseline_KS',
    'exeter_project_mf_aa_KS_30_72,90',
    'exeter_project_mf_baseline_st_KS',
    'exeter_project_mf_aa_st_KS_30_72,90'
]

persistences = []
areas = []
number = []

for exp in experiments:
    dic = np.load(base_path + exp + "/tot/dictionaries/falwa_LWAA_tracked_flipped_2.5x2.5_daily_2-51_DIC.npy",allow_pickle=True)
    number.append(len(dic))
    areas.append([])
    persistences.append([])
    for event in dic[1:]:
        areas[-1].append(sum(event['area'])/event['persistence'])
        persistences[-1].append(event['persistence'])
```

```python
plt.boxplot(areas,showfliers=False)
```

```python
print('durata media')

titles = ['SYM baseline: ','SYM AA:       ','ASYM baseline: ','ASYM AA:       ']

avgs_persistence = []
for i,realisation in enumerate(persistences):
    avgs_persistence.append(sum(realisation)/len(realisation))
    print(titles[i] + str(avgs_persistence[-1]))

print('aumento percentuale:')
print('SYM')
print(((avgs_persistence[1]-avgs_persistence[0])/avgs_persistence[0])*100)
print('ASYM')
print(((avgs_persistence[3]-avgs_persistence[2])/avgs_persistence[2])*100)
```

```python
print('area media')

titles = ['SYM baseline: ','SYM AA:       ','ASYM baseline: ','ASYM AA:       ']

avgs_area = []
for realisation in areas:
    avgs_area.append(sum(realisation)/len(realisation))
    print(avgs_area[-1])
    
print('aumento percentuale:')
print('SYM')
print(((avgs_area[1]-avgs_area[0])/avgs_area[0])*100)
print('ASYM')
print(((avgs_area[3]-avgs_area[0])/avgs_area[0])*100)
```
