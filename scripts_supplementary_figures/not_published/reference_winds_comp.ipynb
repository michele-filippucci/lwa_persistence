```python
import xarray as xr
import matplotlib.pyplot as plt
```

```python
ds_AA_SYM = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_aa_KS_30_72,90/tot/dw_uref_CLIM_2-51.nc')
ds_BASE_SYM = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_baseline_KS/tot/dw_uref_CLIM_2-51.nc')
ds_AA_ASYM = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_aa_st_KS_30_72,90/tot/dw_uref_CLIM_2-51.nc')
ds_BASE_ASYM = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_baseline_st_KS/tot/dw_uref_CLIM_2-51.nc')
```

```python
dw_uref_AA_SYM = ds_AA_SYM['dw_uref'][0,:].loc[0:]
dw_uref_BASE_SYM = ds_BASE_SYM['dw_uref'][0,:].loc[0:]
dw_uref_AA_ASYM = ds_AA_ASYM['dw_uref'][0,:].loc[0:]
dw_uref_BASE_ASYM = ds_BASE_ASYM['dw_uref'][0,:].loc[0:]
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))
 
# --- Plot 1 ---
ax1.plot(dw_uref_AA_SYM['lat'],dw_uref_AA_SYM,label='AA')
ax1.plot(dw_uref_BASE_SYM['lat'],dw_uref_BASE_SYM,label='base')
ax1.set_title("(a) SYM")
ax1.set_ylabel(r'$U_{ref}$ [m/s]')
ax1.set_xlabel(r'lat [°N]')
ax1.legend()
 
# --- Plot 2 ---
ax2.plot(dw_uref_AA_ASYM['lat'],dw_uref_AA_ASYM,label='AA')
ax2.plot(dw_uref_BASE_ASYM['lat'],dw_uref_BASE_ASYM,label='base')
ax2.set_title("(b) ASYM")
ax2.set_ylabel(r'$U_{ref}$ [m/s]')
ax2.set_xlabel(r'lat [°N]')
 
#fig.suptitle("2×1 Multiplot", fontsize=14)
plt.tight_layout()
plt.savefig("/home/links/mf671/github/lwa_persistence/supplementary_plots/Figure_S4.png",dpi=300)
```

```python
dw_zgv_AA_SYM = ds_FC_AA_SYM['zgv'].mean('lon')
dw_zgv_BASE_SYM = ds_FC_BASE_SYM['zgv'].mean('lon')
dw_zgv_AA_ASYM = ds_FC_AA_ASYM['zgv'].mean('lon')
dw_zgv_BASE_ASYM = ds_FC_BASE_ASYM['zgv'].mean('lon')
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))
 
# --- Plot 1 ---
ax1.plot(dw_zgv_AA_SYM['lat'],dw_zgv_AA_SYM - dw_uref_AA_SYM,label='AA zgv')
ax1.plot(dw_zgv_BASE_SYM['lat'],dw_zgv_BASE_SYM - dw_uref_BASE_SYM,label='base zgv')
ax1.plot(dw_uref_AA_SYM['lat'],dw_uref_AA_SYM,label='AA uref')
ax1.plot(dw_uref_BASE_SYM['lat'],dw_uref_BASE_SYM,label='base uref')
ax1.set_title("(a) SYM")
ax1.set_ylabel(r'$U_{ref}$ [m/s]')
ax1.set_xlabel(r'lat [°N]')
ax1.legend()
 
# --- Plot 2 ---
ax2.plot(dw_zgv_AA_ASYM['lat'],dw_zgv_AA_ASYM - dw_uref_AA_ASYM,label='AA uref zgv')
ax2.plot(dw_zgv_BASE_ASYM['lat'],dw_zgv_BASE_ASYM - dw_uref_BASE_ASYM,label='base uref zgv')
ax2.plot(dw_uref_AA_ASYM['lat'],dw_uref_AA_ASYM,label='AA')
ax2.plot(dw_uref_BASE_ASYM['lat'],dw_uref_BASE_ASYM,label='base')
ax2.set_title("(b) ASYM")
ax2.set_ylabel(r'$U_{ref}$ [m/s]')
ax2.set_xlabel(r'lat [°N]')
 
#fig.suptitle("2×1 Multiplot", fontsize=14)
plt.tight_layout()
plt.show()
#plt.savefig("/home/links/mf671/github/lwa_persistence/supplementary_plots/Figure_S4.png",dpi=300)
```
