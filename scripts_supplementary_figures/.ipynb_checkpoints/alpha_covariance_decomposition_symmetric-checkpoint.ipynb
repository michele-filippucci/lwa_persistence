# COVARIANCE CHANGE

Here we try to understand what is causing the change in covariance between U and LWA. We find that the change in covariance is mostly negative and it is mostly driven by a change in mean zonal winds.

```python
import xarray as xr
import matplotlib.pyplot as plt
import numpy as np

# --- INPUT ---
# Assume you have xarray DataArrays: lwa and ucomp, with dims ('time','lat','lon')
# ref: reference period (e.g., control)
# pert: perturbed period (e.g., experiment)

pert = xr.open_dataset('/emmy-noether/home/mf671/DATA/isca_data/exeter_project_mf_aa_st_KS_30_72,90/tot/filtered_decomposition_data_flipped_2-51.nc')
ref = xr.open_dataset('/emmy-noether/home/mf671/DATA/isca_data/exeter_project_mf_baseline_st_KS/tot/filtered_decomposition_data_flipped_2-51.nc')

lat = pert['lat']
ucomp_ref = ref['ucomp']# - ref['ucomp'].mean('time')
ucomp_pert = pert['ucomp']#  - pert['ucomp'].mean('time')
cosphi = np.cos(np.deg2rad(lat)).broadcast_like(ucomp_ref)
dw_lwa_ref = -(ref['dw_lwa'])/cosphi
dw_lwa_pert = -(pert['dw_lwa'])/cosphi
var_ref = dw_lwa_ref.var(dim='time')
var_pert = dw_lwa_pert.var(dim='time')

# --- 1. Means ---
E_X_ref = ucomp_ref.mean(dim='time')# * np.sqrt(dw_lwa_var_ref)
E_Y_ref = dw_lwa_ref.mean(dim='time')# * np.sqrt(dw_lwa_var_ref)
E_X_pert = ucomp_pert.mean(dim='time')# * np.sqrt(dw_lwa_var_pert)
E_Y_pert = dw_lwa_pert.mean(dim='time')# * np.sqrt(dw_lwa_var_pert)

# --- 2. Raw joint moments ---
E_XY_ref = (ucomp_ref*dw_lwa_ref).mean(dim='time')# * dw_lwa_var_ref
E_XY_pert = (ucomp_pert*dw_lwa_pert).mean(dim='time')# * dw_lwa_var_pert

# --- 3. Covariance decomposition ---
dCov_joint = E_XY_pert - E_XY_ref
dCov_Xmean = - E_Y_ref * (E_X_pert - E_X_ref)
dCov_Ymean = - E_X_ref * (E_Y_pert - E_Y_ref)
dCov_interaction = - (E_X_pert - E_X_ref) * (E_Y_pert - E_Y_ref)

# total decomposed change
dCov_total = dCov_joint + dCov_Xmean + dCov_Ymean + dCov_interaction

# --- 4. Actual covariance values ---
cov_ref = E_XY_ref - E_X_ref*E_Y_ref
cov_pert = E_XY_pert - E_X_pert*E_Y_pert
delta_cov = cov_pert - cov_ref
```

```python
#plot and sanity check

# --- 5. Sanity check ---
print("Max difference (should be ~0):", float((delta_cov - dCov_total).max()))

levels = np.arange(-600,600.2,40)/40

# --- 6. Plotting ---
terms = {
    'ΔCov joint': dCov_joint,
    'ΔCov ucomp mean': dCov_Xmean,
    'ΔCov dw_lwa mean': dCov_Ymean,
    'ΔCov interaction': dCov_interaction,
    'ΔCov total': dCov_total,
    'ΔCov actual': delta_cov
}

fig, axes = plt.subplots(2,3, figsize=(18,10), constrained_layout=True)
axes = axes.flatten()

for ax, (name, da) in zip(axes, terms.items()):
    im = da.plot(ax=ax, levels= levels, cmap='bwr', add_colorbar=True)
    ax.set_title(name)

plt.show()
```

# $\beta$ changes

Here we combine the informations from above with the changes in the variances of LWA, which contribute to estimating $\alpha$, the regression coefficient between LWA and U.  
We find that the midlatitude LWA variance increase is even more important in shaping the ALPHA response to arctic amplification.  
On the other hand, the decrease of LWA variance at high latitudes compensate the decrease of covariance of U and LWA, resulting in a net zero effect on $\alpha$

```python
# First we plot the change of variance:

# --- Plot maps ---
import matplotlib.pyplot as plt
terms = {
    'Δvar': var_pert-var_ref,
}

fig, axes = plt.subplots(1,1, figsize=(8,5), constrained_layout=True)

#levels = np.arange()
im = (var_pert-var_ref).plot(ax=axes, cmap='bwr', center=0, add_colorbar=True)
axes.set_title('Δvar')

plt.show()
```

```python

# Assume you already have:
# cov_ref, cov_pert, var_ref, var_pert  (all xarray DataArray, dims lat/lon)
# derived earlier as:
# cov_ref  = E_XY_ref  - E_X_ref * E_Y_ref
# cov_pert = E_XY_pert - E_X_pert * E_Y_pert
# var_ref, var_pert = lwa_ref.var(dim='time'), lwa_pert.var(dim='time')

# Delta cov and delta beta actual
dC = cov_pert - cov_ref
beta_ref  = cov_ref  / var_ref
beta_pert = cov_pert / var_pert
dBeta_actual = beta_pert - beta_ref

# --- Decomposition A (denominator = var_pert) ---
dBeta_cov_A = dC / var_pert                      # ΔC / Vp
dBeta_var_A = cov_ref * (1/var_pert - 1/var_ref) # Cr*(1/Vp - 1/Vr)
dBeta_A_sum = dBeta_cov_A + dBeta_var_A          # should equal dBeta_actual (within numerical tolerance)

# --- Decomposition B (denominator = var_ref) ---
dBeta_cov_B = dC / var_ref                       # ΔC / Vr
dBeta_var_B = cov_pert * (1/var_pert - 1/var_ref) # Cp*(1/Vp - 1/Vr)
dBeta_B_sum = dBeta_cov_B + dBeta_var_B

# Sanity checks (small numerical residuals)
#print("Max residual A:", float((dBeta_actual - dBeta_A_sum).abs().max()))
#print("Max residual B:", float((dBeta_actual - dBeta_B_sum).abs().max()))

# Optional: fraction contributions (pointwise)
frac_cov_A = dBeta_cov_A / dBeta_actual
frac_var_A = dBeta_var_A / dBeta_actual

# --- Plot maps ---
import matplotlib.pyplot as plt
terms = {
    'Δβ actual': dBeta_actual,
    'Δβ cov (ΔC/Vp)': dBeta_cov_A,
    'Δβ var (Cr*(1/Vp-1/Vr))': dBeta_var_A,
    'Δβ sum (A)': dBeta_A_sum,
}

fig, axes = plt.subplots(2,2, figsize=(12,8), constrained_layout=True)
axes = axes.ravel()

levels = np.arange(-0.055,0.06,0.01)

for ax, (name, da) in zip(axes, terms.items()):
    im = da.plot(ax=ax, cmap='bwr', levels=levels, add_colorbar=True)
    ax.set_title(name)
plt.show()
```
