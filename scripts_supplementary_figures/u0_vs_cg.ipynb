```python
import xarray as xr
import matplotlib.pyplot as plt
```

```python
ds_base = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_baseline_KS/tot/filtered_decomposition_data_flipped_2-51.nc')
ds2_base = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_baseline_KS/tot/FC_1.5x1.5_2-51_wbeta.nc')

ds_aa = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_aa_KS_30_72,90/tot/filtered_decomposition_data_flipped_2-51.nc')
ds2_aa = xr.open_dataset('/home/links/mf671/DATA/isca_data/exeter_project_mf_aa_KS_30_72,90/tot/FC_1.5x1.5_2-51_wbeta.nc')
```

```python
ds2_base['zgv'].shape
```

```python
uref_base = ds_base['dw_uref'].mean('time')
zgv_base = ds2_base['zgv'].mean('lon')

uref_aa = ds_aa['dw_uref'].mean('time')
zgv_aa = ds2_aa['zgv'].mean('lon')
```

```python
plt.plot(uref_base.lat,uref_base,color='blue', label = 'time-mean reference wind baseline')
plt.plot(uref_aa.lat,uref_aa,color='cyan', label = 'time-mean reference wind aa')
plt.plot(uref_base.lat,zgv_base-uref_base,color='red',label = 'zonal group velocity baseline')
plt.plot(uref_aa.lat,zgv_aa-uref_aa,color='orange',label = 'zonal group velocity aa')
plt.ylabel('velocity (m/s)')
plt.xlabel('latitude (°N)')
plt.legend()
plt.ylim(-10,20)
```
