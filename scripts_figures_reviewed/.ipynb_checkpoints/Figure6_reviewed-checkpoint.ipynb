```python
def compute_FC(zon_group_velocity, u_0, alpha, beta, density_weighted_stat_lwa):

    lat = zon_group_velocity['lat']
    cosphi = np.cos(np.deg2rad(lat)).broadcast_like(zon_group_velocity)
    
    # intermediate
    u = zon_group_velocity + u_0 + beta - 2*alpha*density_weighted_stat_lwa
    
    #computed FC
    FC = (u**2 * cosphi) / (4.0 * alpha)
    return FC

```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import cartopy.crs as ccrs

plt.rcParams['font.size'] = 12

# --------------------------------------------------
# Helper
# --------------------------------------------------
def broadcast_zonal(field, ds_ref):
    return field.mean('lon').broadcast_like(ds_ref)

# --------------------------------------------------
# Figure layout
# --------------------------------------------------
fig = plt.figure(figsize=(17,17))

gs = gridspec.GridSpec(
    nrows=5,
    ncols=5,
    width_ratios=[6.0,0.25,1.2,0.01,6.0],
    hspace=0.45,
    wspace=-0.3
)

titles  = ['symmetric','asymmetric']
letters = ['(a)','(b)','(c)','(d)','(e)','(f)','(g)','(h)','(i)','(l)']
colors  = ['k','silver']

coords = [(-75.0,25.0),(-50.0,25.0),(-50.0,50.0)]
lons,lats = zip(*coords)
lons = np.array(lons)
lats = np.array(lats)

lons_closed = np.append(lons,lons[0])
lats_closed = np.append(lats,lats[0])

proj = ccrs.PlateCarree()

levels = np.arange(40,400,40)*3
levels_diff = np.arange(-42.75,43,1.5)

ax_xsec = []

# --------------------------------------------------
# Loop symmetric / asymmetric
# --------------------------------------------------
for i in range(2):

    XSEC_COL = 2
    MAP_COL  = 0 if i==0 else 4

    ds_a = ds_clims[i*2]
    ds_b = ds_clims[i*2+1]

    # ==================================================
    # ROW 0 — TOTAL DIFFERENCE (moved to top)
    # ==================================================
    if i==0:
        ax_xsec_total = fig.add_subplot(gs[0, XSEC_COL])

    ax_map = fig.add_subplot(gs[0, MAP_COL], projection=proj, sharey=ax_xsec_total)

    FC_b = compute_FC(ds_b['zgv']-ds_b['u_0'],ds_b['u_0'],ds_b['alpha'],ds_b['beta'],ds_b['qslwa'])
    FC_a = compute_FC(ds_a['zgv']-ds_a['u_0'],ds_a['u_0'],ds_a['alpha'],ds_a['beta'],ds_a['qslwa'])

    if i==0:
        FC_b = broadcast_zonal(FC_b,ds_a)
        FC_a = broadcast_zonal(FC_a,ds_a)

    total_diff = FC_b - FC_a

    ax_map.set_title(f'{letters[i]} Total difference – {titles[i]}')

    c1 = ax_map.contour(ds_a.lon,ds_a.lat,FC_a,
                        levels=levels,colors='dimgray')

    ax_map.contour(ds_b.lon,ds_b.lat,FC_b,
                   levels=levels,colors='dimgray',
                   linestyles='dashed')

    ax_map.clabel(c1,inline=True,colors='black',fontsize=11)

    sh = ax_map.contourf(ds_a.lon,ds_a.lat,total_diff,
                         levels=levels_diff,
                         cmap='RdBu_r',extend='both')

    if i>0:
        fig.colorbar(sh,ax=ax_map,label=r'$\Delta F_c$ [W m$^{-2}$]')

    if i>0:
        ax_map.plot(lons_closed,lats_closed,'--',lw=3,color='red',transform=proj)

    ax_map.set_extent([ds_a.lon.min(),ds_a.lon.max(),10,85])
    ax_map.set_aspect(2.5)

    ax_map.set_xticks([-120,-60,0,60,120],crs=proj)
    ax_map.set_xticklabels(['120°W','60°W','0°E','60°E','120°E'])
    ax_map.set_xlabel('Longitude')

    if i==0:
        ax_map.set_yticks([20,40,60,80])
        ax_map.set_yticklabels(['20°N','40°N','60°N','80°N'])
        ax_map.set_ylabel('Latitude')
    else:
        plt.setp(ax_map.get_yticklabels(),visible=False)

    ax_map.gridlines(draw_labels=False,linestyle='--',alpha=0.6)

    xsec = total_diff.mean('lon').values

    ax_xsec_total.plot(xsec,ds_a.lat.values,color=colors[i],lw=2,label=titles[i])
    ax_xsec_total.axvline(0,color='gray',ls='--')
    ax_xsec_total.set_xlim(-35,35)
    ax_xsec_total.set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
    ax_xsec_total.grid(True,ls='--',alpha=0.5)
    ax_xsec_total.tick_params(axis='y',length=0)
    plt.setp(ax_xsec_total.get_yticklabels(),visible=False)

    if i==1:
        ax_xsec_total.legend(loc='upper right',bbox_to_anchor=(1.0,1.35))

    # ==================================================
    # FC_terms definitions
    # ==================================================
    FC_terms = [
        ('Alpha and beta contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'],ds_a['u_0'],ds_b['alpha'],ds_b['beta'],ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'],ds_a['u_0'],ds_a['alpha'],ds_a['beta'],ds_a['qslwa'])),

        ('Reference wind contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'],ds_b['u_0'],ds_a['alpha'],ds_a['beta'],ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'],ds_a['u_0'],ds_a['alpha'],ds_a['beta'],ds_a['qslwa'])),

        ('Zonal group velocity contribution',
         compute_FC(ds_b['zgv']-ds_b['u_0'],ds_a['u_0'],ds_a['alpha'],ds_a['beta'],ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'],ds_a['u_0'],ds_a['alpha'],ds_a['beta'],ds_a['qslwa'])),

        ('Stationary LWA contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'],ds_a['u_0'],ds_a['alpha'],ds_a['beta'],ds_b['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'],ds_a['u_0'],ds_a['alpha'],ds_a['beta'],ds_a['qslwa']))
    ]

    # ==================================================
    # ROWS 1–3 — linear terms
    # ==================================================
    for r,(label,field_b,field_a) in enumerate(FC_terms):

        if i==0:
            ax_xsec.append(fig.add_subplot(gs[r+1, XSEC_COL]))

        ax_map = fig.add_subplot(gs[r+1, MAP_COL], projection=proj, sharey=ax_xsec[r])

        if i==0:
            field_b = broadcast_zonal(field_b,ds_a)
            field_a = broadcast_zonal(field_a,ds_a)

        fc_diff = field_b - field_a

        ax_map.set_title(f'{letters[2*(r+1) + i]} {label} – {titles[i]}')

        c1 = ax_map.contour(ds_a.lon,ds_a.lat,field_a,
                            levels=levels,colors='dimgray')

        ax_map.contour(ds_b.lon,ds_b.lat,field_b,
                       levels=levels,colors='dimgray',
                       linestyles='dashed')

        ax_map.clabel(c1,inline=True,colors='black',fontsize=11)

        sh = ax_map.contourf(ds_a.lon,ds_a.lat,fc_diff,
                             levels=levels_diff,
                             cmap='RdBu_r',extend='both')

        if i>0:
            fig.colorbar(sh,ax=ax_map,label=r'$\Delta F_c$ [W m$^{-2}$]')

        if i>0:
            ax_map.plot(lons_closed,lats_closed,'--',lw=3,color='red',transform=proj)

        ax_map.set_extent([ds_a.lon.min(),ds_a.lon.max(),10,85])
        ax_map.set_aspect(2.5)

        ax_map.set_xticks([-120,-60,0,60,120],crs=proj)
        ax_map.set_xticklabels(['120°W','60°W','0°E','60°E','120°E'])
        ax_map.set_xlabel('Longitude')

        if i==0:
            ax_map.set_yticks([20,40,60,80])
            ax_map.set_yticklabels(['20°N','40°N','60°N','80°N'])
            ax_map.set_ylabel('Latitude')
        else:
            plt.setp(ax_map.get_yticklabels(),visible=False)

        ax_map.gridlines(draw_labels=False,linestyle='--',alpha=0.6)

        xsec = fc_diff.mean('lon').values

        if r==3 and i==0:
            xsec = np.zeros_like(xsec)

        ax_xsec[r].plot(xsec,ds_a.lat.values,color=colors[i],lw=2,label=titles[i])
        ax_xsec[r].axvline(0,color='gray',ls='--')
        ax_xsec[r].set_xlim(-35,35)
        ax_xsec[r].set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
        ax_xsec[r].grid(True,ls='--',alpha=0.5)
        ax_xsec[r].tick_params(axis='y',length=0)
        plt.setp(ax_xsec[r].get_yticklabels(),visible=False)

# --------------------------------------------------
plt.savefig('PLOTS/Figure7_refactored.png',dpi=500,bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import cartopy.crs as ccrs

plt.rcParams['font.size'] = 12
plt.rcParams['hatch.color'] = 'k'
plt.rcParams['hatch.linewidth'] = 0.6

# ==================================================
# Settings for significance
# ==================================================
Z_SIG = 1.96   # 95% two-sided
N_EFF = 1      # if *_std is the interannual spread, use number of years/members
               # (std of the mean = std/sqrt(N_EFF)); keep 1 if *_std is already
               # the uncertainty of the mean

# dataset ('A' or 'B') feeding each slot of compute_FC: (c, u_0, alpha, beta, qslwa)
SPEC = {
    'total': ('B', 'B', 'B', 'B', 'B'),
    'ab':    ('A', 'A', 'B', 'B', 'A'),
    'u0':    ('A', 'B', 'A', 'A', 'A'),
    'cg':    ('B', 'A', 'A', 'A', 'A'),
    'q':     ('A', 'A', 'A', 'A', 'B'),
}
BASE = ('A',) * 5


# ==================================================
# Helpers
# ==================================================
def broadcast_zonal(field, ds_ref):
    return field.mean('lon').broadcast_like(ds_ref)


def _grad_coefs(dsets, spec, lat):
    """dF/d(elementary variable), keyed by (dataset, variable).
    Elementary variables: zgv, u_0, alpha, beta (c = zgv - u_0)."""
    sc, su, sa, sb, sq = spec
    c  = dsets[sc]['zgv'] - dsets[sc]['u_0']
    u0 = dsets[su]['u_0']
    al = dsets[sa]['alpha']
    be = dsets[sb]['beta']
    q  = dsets[sq]['qslwa']
    cosphi = np.cos(np.deg2rad(lat))

    u  = c + u0 + be - 2*al*q
    Fu = u*cosphi/(2*al)                              # dF/du
    Fa = -u**2*cosphi/(4*al**2) - 2*q*Fu              # dF/dalpha

    coef = {}
    def add(key, val):
        coef[key] = coef.get(key, 0) + val

    add((sc, 'zgv'),   Fu)    # through c = zgv - u0
    add((sc, 'u_0'),  -Fu)
    add((su, 'u_0'),   Fu)    # explicit u_0 slot
    add((sa, 'alpha'), Fa)
    add((sb, 'beta'),  Fu)
    return coef


def propagate_sigma(ds_a, ds_b, zonal):
    """Returns {term: (sigma_map[lat,lon], sigma_xsec[lat])} of the plotted difference."""
    ds   = {'A': ds_a, 'B': ds_b}
    lat  = ds_a['lat']
    nlon = ds_a.sizes['lon']
    ref  = ds_a['zgv']
    base = _grad_coefs(ds, BASE, lat)

    out = {}
    for k, spec in SPEC.items():
        g = _grad_coefs(ds, spec, lat)
        var = 0
        for key in set(g) | set(base):
            coef  = g.get(key, 0) - base.get(key, 0)
            sigma = ds[key[0]][key[1] + '_std'] / np.sqrt(N_EFF)
            var   = var + (coef*sigma)**2
        var = var.broadcast_like(ref)

        # zonal mean of the difference, errors independent between longitudes
        sig_x = np.sqrt(var.mean('lon') / nlon)
        sig_m = sig_x.broadcast_like(ref) if zonal else np.sqrt(var)
        out[k] = (sig_m, sig_x)
    return out


def add_hatch(ax, lon, lat, diff, sigma, hatch='...', where='significant'):
    """Hatch non-significant points (default) or significant ones (where='significant')."""
    sig_mask = np.abs(diff) > Z_SIG*sigma
    if where == 'not_significant':
        # NaNs in diff/sigma are not hatched
        mask = (~sig_mask & diff.notnull() & sigma.notnull()).astype(float).values
    else:
        mask = sig_mask.astype(float).fillna(0).values
    if mask.any():
        ax.contourf(lon, lat, mask, levels=[0.5, 1.5], colors='none',
                    hatches=[hatch], transform=ccrs.PlateCarree())

# ==================================================
# Figure layout
# ==================================================
fig = plt.figure(figsize=(17, 17))

gs = gridspec.GridSpec(
    nrows=5,
    ncols=5,
    width_ratios=[6.0, 0.25, 1.2, 0.01, 6.0],
    hspace=0.45,
    wspace=-0.3
)

titles  = ['symmetric', 'asymmetric']
letters = ['(a)', '(b)', '(c)', '(d)', '(e)', '(f)', '(g)', '(h)', '(i)', '(l)']
colors  = ['k', 'silver']

coords = [(-75.0, 25.0), (-50.0, 25.0), (-50.0, 50.0)]
lons, lats = zip(*coords)
lons = np.array(lons)
lats = np.array(lats)

lons_closed = np.append(lons, lons[0])
lats_closed = np.append(lats, lats[0])

proj = ccrs.PlateCarree()

levels = np.arange(40, 400, 40)*3
levels_diff = np.arange(-42.75, 43, 1.5)

ax_xsec = []
term_keys = ['ab', 'u0', 'cg', 'q']

# ==================================================
# Loop symmetric / asymmetric
# ==================================================
for i in range(2):

    XSEC_COL = 2
    MAP_COL  = 0 if i == 0 else 4

    ds_a = ds_clims[i*2]
    ds_b = ds_clims[i*2+1]

    # error propagation for all terms of this case
    sig = propagate_sigma(ds_a, ds_b, zonal=(i == 0))

    # ==================================================
    # ROW 0 — TOTAL DIFFERENCE
    # ==================================================
    if i == 0:
        ax_xsec_total = fig.add_subplot(gs[0, XSEC_COL])

    ax_map = fig.add_subplot(gs[0, MAP_COL], projection=proj, sharey=ax_xsec_total)

    FC_b = compute_FC(ds_b['zgv']-ds_b['u_0'], ds_b['u_0'], ds_b['alpha'], ds_b['beta'], ds_b['qslwa'])
    FC_a = compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])

    if i == 0:
        FC_b = broadcast_zonal(FC_b, ds_a)
        FC_a = broadcast_zonal(FC_a, ds_a)

    total_diff = FC_b - FC_a

    ax_map.set_title(f'{letters[i]} Total difference – {titles[i]}')

    c1 = ax_map.contour(ds_a.lon, ds_a.lat, FC_a,
                        levels=levels, colors='dimgray')

    ax_map.contour(ds_b.lon, ds_b.lat, FC_b,
                   levels=levels, colors='dimgray',
                   linestyles='dashed')

    ax_map.clabel(c1, inline=True, colors='black', fontsize=11)

    sh = ax_map.contourf(ds_a.lon, ds_a.lat, total_diff,
                         levels=levels_diff,
                         cmap='RdBu_r', extend='both')

    # hatching: |dFc| > Z_SIG * sigma
    add_hatch(ax_map, ds_a.lon, ds_a.lat, total_diff, sig['total'][0])

    if i > 0:
        fig.colorbar(sh, ax=ax_map, label=r'$\Delta F_c$ [W m$^{-2}$]')

    if i > 0:
        ax_map.plot(lons_closed, lats_closed, '--', lw=3, color='red', transform=proj)

    ax_map.set_extent([ds_a.lon.min(), ds_a.lon.max(), 10, 85])
    ax_map.set_aspect(2.5)

    ax_map.set_xticks([-120, -60, 0, 60, 120], crs=proj)
    ax_map.set_xticklabels(['120°W', '60°W', '0°E', '60°E', '120°E'])
    ax_map.set_xlabel('Longitude')

    if i == 0:
        ax_map.set_yticks([20, 40, 60, 80])
        ax_map.set_yticklabels(['20°N', '40°N', '60°N', '80°N'])
        ax_map.set_ylabel('Latitude')
    else:
        plt.setp(ax_map.get_yticklabels(), visible=False)

    ax_map.gridlines(draw_labels=False, linestyle='--', alpha=0.6)

    xsec = total_diff.mean('lon').values

    ax_xsec_total.plot(xsec, ds_a.lat.values, color=colors[i], lw=2, label=titles[i])
    sx = sig['total'][1].values
    ax_xsec_total.fill_betweenx(ds_a.lat.values, xsec - Z_SIG*sx, xsec + Z_SIG*sx,
                                color=colors[i], alpha=0.25, lw=0)
    ax_xsec_total.axvline(0, color='gray', ls='--')
    ax_xsec_total.set_xlim(-35, 35)
    ax_xsec_total.set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
    ax_xsec_total.grid(True, ls='--', alpha=0.5)
    ax_xsec_total.tick_params(axis='y', length=0)
    plt.setp(ax_xsec_total.get_yticklabels(), visible=False)

    if i == 1:
        ax_xsec_total.legend(loc='upper right', bbox_to_anchor=(1.0, 1.35))

    # ==================================================
    # FC_terms definitions
    # ==================================================
    FC_terms = [
        ('Alpha and beta contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_b['alpha'], ds_b['beta'], ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])),

        ('Reference wind contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_b['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])),

        ('Zonal group velocity contribution',
         compute_FC(ds_b['zgv']-ds_b['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])),

        ('Stationary LWA contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_b['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']))
    ]

    # ==================================================
    # ROWS 1–4 — linear terms
    # ==================================================
    for r, (label, field_b, field_a) in enumerate(FC_terms):

        key = term_keys[r]

        if i == 0:
            ax_xsec.append(fig.add_subplot(gs[r+1, XSEC_COL]))

        ax_map = fig.add_subplot(gs[r+1, MAP_COL], projection=proj, sharey=ax_xsec[r])

        if i == 0:
            field_b = broadcast_zonal(field_b, ds_a)
            field_a = broadcast_zonal(field_a, ds_a)

        fc_diff = field_b - field_a

        ax_map.set_title(f'{letters[2*(r+1) + i]} {label} – {titles[i]}')

        c1 = ax_map.contour(ds_a.lon, ds_a.lat, field_a,
                            levels=levels, colors='dimgray')

        ax_map.contour(ds_b.lon, ds_b.lat, field_b,
                       levels=levels, colors='dimgray',
                       linestyles='dashed')

        ax_map.clabel(c1, inline=True, colors='black', fontsize=11)

        sh = ax_map.contourf(ds_a.lon, ds_a.lat, fc_diff,
                             levels=levels_diff,
                             cmap='RdBu_r', extend='both')

        # hatching: |dFc| > Z_SIG * sigma
        add_hatch(ax_map, ds_a.lon, ds_a.lat, fc_diff, sig[key][0])

        if i > 0:
            fig.colorbar(sh, ax=ax_map, label=r'$\Delta F_c$ [W m$^{-2}$]')

        if i > 0:
            ax_map.plot(lons_closed, lats_closed, '--', lw=3, color='red', transform=proj)

        ax_map.set_extent([ds_a.lon.min(), ds_a.lon.max(), 10, 85])
        ax_map.set_aspect(2.5)

        ax_map.set_xticks([-120, -60, 0, 60, 120], crs=proj)
        ax_map.set_xticklabels(['120°W', '60°W', '0°E', '60°E', '120°E'])
        ax_map.set_xlabel('Longitude')

        if i == 0:
            ax_map.set_yticks([20, 40, 60, 80])
            ax_map.set_yticklabels(['20°N', '40°N', '60°N', '80°N'])
            ax_map.set_ylabel('Latitude')
        else:
            plt.setp(ax_map.get_yticklabels(), visible=False)

        ax_map.gridlines(draw_labels=False, linestyle='--', alpha=0.6)

        xsec = fc_diff.mean('lon').values

        if r == 3 and i == 0:
            xsec = np.zeros_like(xsec)

        ax_xsec[r].plot(xsec, ds_a.lat.values, color=colors[i], lw=2, label=titles[i])
        if not (r == 3 and i == 0):
            sx = sig[key][1].values
            ax_xsec[r].fill_betweenx(ds_a.lat.values, xsec - Z_SIG*sx, xsec + Z_SIG*sx,
                                     color=colors[i], alpha=0.25, lw=0)
        ax_xsec[r].axvline(0, color='gray', ls='--')
        ax_xsec[r].set_xlim(-35, 35)
        ax_xsec[r].set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
        ax_xsec[r].grid(True, ls='--', alpha=0.5)
        ax_xsec[r].tick_params(axis='y', length=0)
        plt.setp(ax_xsec[r].get_yticklabels(), visible=False)

# --------------------------------------------------
plt.savefig('PLOTS/Figure7_refactored.png', dpi=500, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import cartopy.crs as ccrs

plt.rcParams['font.size'] = 12
plt.rcParams['hatch.color'] = 'k'
plt.rcParams['hatch.linewidth'] = 0.6

# ==================================================
# Settings for significance
# ==================================================
Z_SIG = 1.96   # 95% two-sided
N_EFF = 1      # if *_std is the interannual spread, use number of years/members
               # (std of the mean = std/sqrt(N_EFF)); keep 1 if *_std is already
               # the uncertainty of the mean

# dataset ('A' or 'B') feeding each slot of compute_FC: (c, u_0, alpha, beta, qslwa)
SPEC = {
    'total': ('B', 'B', 'B', 'B', 'B'),
    'al':    ('A', 'A', 'B', 'A', 'A'),   # alpha only
    'be':    ('A', 'A', 'A', 'B', 'A'),   # beta only
    'u0':    ('A', 'B', 'A', 'A', 'A'),
    'cg':    ('B', 'A', 'A', 'A', 'A'),
    'q':     ('A', 'A', 'A', 'A', 'B'),
}
BASE = ('A',) * 5


# ==================================================
# Helpers
# ==================================================
def broadcast_zonal(field, ds_ref):
    return field.mean('lon').broadcast_like(ds_ref)


def _grad_coefs(dsets, spec, lat):
    """dF/d(elementary variable), keyed by (dataset, variable).
    Elementary variables: zgv, u_0, alpha, beta (c = zgv - u_0)."""
    sc, su, sa, sb, sq = spec
    c  = dsets[sc]['zgv'] - dsets[sc]['u_0']
    u0 = dsets[su]['u_0']
    al = dsets[sa]['alpha']
    be = dsets[sb]['beta']
    q  = dsets[sq]['qslwa']
    cosphi = np.cos(np.deg2rad(lat))

    u  = c + u0 + be - 2*al*q
    Fu = u*cosphi/(2*al)                              # dF/du
    Fa = -u**2*cosphi/(4*al**2) - 2*q*Fu              # dF/dalpha

    coef = {}
    def add(key, val):
        coef[key] = coef.get(key, 0) + val

    add((sc, 'zgv'),   Fu)    # through c = zgv - u0
    add((sc, 'u_0'),  -Fu)
    add((su, 'u_0'),   Fu)    # explicit u_0 slot
    add((sa, 'alpha'), Fa)
    add((sb, 'beta'),  Fu)
    return coef


def propagate_sigma(ds_a, ds_b, zonal):
    """Returns {term: (sigma_map[lat,lon], sigma_xsec[lat])} of the plotted difference."""
    ds   = {'A': ds_a, 'B': ds_b}
    lat  = ds_a['lat']
    nlon = ds_a.sizes['lon']
    ref  = ds_a['zgv']
    base = _grad_coefs(ds, BASE, lat)

    out = {}
    for k, spec in SPEC.items():
        g = _grad_coefs(ds, spec, lat)
        var = 0
        for key in set(g) | set(base):
            coef  = g.get(key, 0) - base.get(key, 0)
            sigma = ds[key[0]][key[1] + '_std'] / np.sqrt(N_EFF)
            var   = var + (coef*sigma)**2
        var = var.broadcast_like(ref)

        # zonal mean of the difference, errors independent between longitudes
        sig_x = np.sqrt(var.mean('lon') / nlon)
        sig_m = sig_x.broadcast_like(ref) if zonal else np.sqrt(var)
        out[k] = (sig_m, sig_x)
    return out


def add_hatch(ax, lon, lat, diff, sigma, hatch='...', where='significant'):
    """Hatch non-significant points (default) or significant ones (where='significant')."""
    sig_mask = np.abs(diff) > Z_SIG*sigma
    if where == 'not_significant':
        # NaNs in diff/sigma are not hatched
        mask = (~sig_mask & diff.notnull() & sigma.notnull()).astype(float).values
    else:
        mask = sig_mask.astype(float).fillna(0).values
    if mask.any():
        ax.contourf(lon, lat, mask, levels=[0.5, 1.5], colors='none',
                    hatches=[hatch], transform=ccrs.PlateCarree())

# ==================================================
# Figure layout
# ==================================================
fig = plt.figure(figsize=(17, 20))

gs = gridspec.GridSpec(
    nrows=6,
    ncols=5,
    width_ratios=[6.0, 0.25, 1.2, 0.01, 6.0],
    hspace=0.45,
    wspace=-0.3
)

titles  = ['symmetric', 'asymmetric']
# 6 rows x 2 columns = 12 panel letters (Italian alphabet: ..., i, l, m, n)
letters = ['(a)', '(b)', '(c)', '(d)', '(e)', '(f)', '(g)', '(h)', '(i)', '(l)', '(m)', '(n)']
colors  = ['k', 'silver']

coords = [(-75.0, 25.0), (-50.0, 25.0), (-50.0, 50.0)]
lons, lats = zip(*coords)
lons = np.array(lons)
lats = np.array(lats)

lons_closed = np.append(lons, lons[0])
lats_closed = np.append(lats, lats[0])

proj = ccrs.PlateCarree()

levels = np.arange(40, 400, 40)*3
levels_diff = np.arange(-42.75, 43, 1.5)

ax_xsec = []
term_keys = ['al', 'be', 'u0', 'cg', 'q']

# ==================================================
# Loop symmetric / asymmetric
# ==================================================
for i in range(2):

    XSEC_COL = 2
    MAP_COL  = 0 if i == 0 else 4

    ds_a = ds_clims[i*2]
    ds_b = ds_clims[i*2+1]

    # error propagation for all terms of this case
    sig = propagate_sigma(ds_a, ds_b, zonal=(i == 0))

    # ==================================================
    # ROW 0 — TOTAL DIFFERENCE
    # ==================================================
    if i == 0:
        ax_xsec_total = fig.add_subplot(gs[0, XSEC_COL])

    ax_map = fig.add_subplot(gs[0, MAP_COL], projection=proj, sharey=ax_xsec_total)

    FC_b = compute_FC(ds_b['zgv']-ds_b['u_0'], ds_b['u_0'], ds_b['alpha'], ds_b['beta'], ds_b['qslwa'])
    FC_a = compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])

    if i == 0:
        FC_b = broadcast_zonal(FC_b, ds_a)
        FC_a = broadcast_zonal(FC_a, ds_a)

    total_diff = FC_b - FC_a

    ax_map.set_title(f'{letters[i]} Total difference – {titles[i]}')

    c1 = ax_map.contour(ds_a.lon, ds_a.lat, FC_a,
                        levels=levels, colors='dimgray')

    ax_map.contour(ds_b.lon, ds_b.lat, FC_b,
                   levels=levels, colors='dimgray',
                   linestyles='dashed')

    ax_map.clabel(c1, inline=True, colors='black', fontsize=11)

    sh = ax_map.contourf(ds_a.lon, ds_a.lat, total_diff,
                         levels=levels_diff,
                         cmap='RdBu_r', extend='both')

    # hatching: |dFc| > Z_SIG * sigma
    add_hatch(ax_map, ds_a.lon, ds_a.lat, total_diff, sig['total'][0])

    if i > 0:
        fig.colorbar(sh, ax=ax_map, label=r'$\Delta F_c$ [W m$^{-2}$]')

    if i > 0:
        ax_map.plot(lons_closed, lats_closed, '--', lw=3, color='red', transform=proj)

    ax_map.set_extent([ds_a.lon.min(), ds_a.lon.max(), 10, 85])
    ax_map.set_aspect(2.5)

    ax_map.set_xticks([-120, -60, 0, 60, 120], crs=proj)
    ax_map.set_xticklabels(['120°W', '60°W', '0°E', '60°E', '120°E'])
    ax_map.set_xlabel('Longitude')

    if i == 0:
        ax_map.set_yticks([20, 40, 60, 80])
        ax_map.set_yticklabels(['20°N', '40°N', '60°N', '80°N'])
        ax_map.set_ylabel('Latitude')
    else:
        plt.setp(ax_map.get_yticklabels(), visible=False)

    ax_map.gridlines(draw_labels=False, linestyle='--', alpha=0.6)

    xsec = total_diff.mean('lon').values

    ax_xsec_total.plot(xsec, ds_a.lat.values, color=colors[i], lw=2, label=titles[i])
    sx = sig['total'][1].values
    ax_xsec_total.fill_betweenx(ds_a.lat.values, xsec - Z_SIG*sx, xsec + Z_SIG*sx,
                                color=colors[i], alpha=0.25, lw=0)
    ax_xsec_total.axvline(0, color='gray', ls='--')
    ax_xsec_total.set_xlim(-35, 35)
    ax_xsec_total.set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
    ax_xsec_total.grid(True, ls='--', alpha=0.5)
    ax_xsec_total.tick_params(axis='y', length=0)
    plt.setp(ax_xsec_total.get_yticklabels(), visible=False)

    if i == 1:
        ax_xsec_total.legend(loc='upper right', bbox_to_anchor=(1.0, 1.35))

    # ==================================================
    # FC_terms definitions (order must match term_keys)
    # ==================================================
    FC_base = compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])

    FC_terms = [
        ('Alpha contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_b['alpha'], ds_a['beta'], ds_a['qslwa']),
         FC_base),

        ('Beta contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_b['beta'], ds_a['qslwa']),
         FC_base),

        ('Reference wind contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_b['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']),
         FC_base),

        ('Zonal group velocity contribution',
         compute_FC(ds_b['zgv']-ds_b['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']),
         FC_base),

        ('Stationary LWA contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_b['qslwa']),
         FC_base)
    ]

    # ==================================================
    # ROWS 1–5 — linear terms
    # ==================================================
    for r, (label, field_b, field_a) in enumerate(FC_terms):

        key = term_keys[r]

        if i == 0:
            ax_xsec.append(fig.add_subplot(gs[r+1, XSEC_COL]))

        ax_map = fig.add_subplot(gs[r+1, MAP_COL], projection=proj, sharey=ax_xsec[r])

        if i == 0:
            field_b = broadcast_zonal(field_b, ds_a)
            field_a = broadcast_zonal(field_a, ds_a)

        fc_diff = field_b - field_a

        ax_map.set_title(f'{letters[2*(r+1) + i]} {label} – {titles[i]}')

        c1 = ax_map.contour(ds_a.lon, ds_a.lat, field_a,
                            levels=levels, colors='dimgray')

        ax_map.contour(ds_b.lon, ds_b.lat, field_b,
                       levels=levels, colors='dimgray',
                       linestyles='dashed')

        ax_map.clabel(c1, inline=True, colors='black', fontsize=11)

        sh = ax_map.contourf(ds_a.lon, ds_a.lat, fc_diff,
                             levels=levels_diff,
                             cmap='RdBu_r', extend='both')

        # hatching: |dFc| > Z_SIG * sigma
        add_hatch(ax_map, ds_a.lon, ds_a.lat, fc_diff, sig[key][0])

        if i > 0:
            fig.colorbar(sh, ax=ax_map, label=r'$\Delta F_c$ [W m$^{-2}$]')

        if i > 0:
            ax_map.plot(lons_closed, lats_closed, '--', lw=3, color='red', transform=proj)

        ax_map.set_extent([ds_a.lon.min(), ds_a.lon.max(), 10, 85])
        ax_map.set_aspect(2.5)

        ax_map.set_xticks([-120, -60, 0, 60, 120], crs=proj)
        ax_map.set_xticklabels(['120°W', '60°W', '0°E', '60°E', '120°E'])
        ax_map.set_xlabel('Longitude')

        if i == 0:
            ax_map.set_yticks([20, 40, 60, 80])
            ax_map.set_yticklabels(['20°N', '40°N', '60°N', '80°N'])
            ax_map.set_ylabel('Latitude')
        else:
            plt.setp(ax_map.get_yticklabels(), visible=False)

        ax_map.gridlines(draw_labels=False, linestyle='--', alpha=0.6)

        xsec = fc_diff.mean('lon').values

        # stationary-LWA term is zero in the symmetric case by construction
        skip_band = (key == 'q' and i == 0)
        if skip_band:
            xsec = np.zeros_like(xsec)

        ax_xsec[r].plot(xsec, ds_a.lat.values, color=colors[i], lw=2, label=titles[i])
        if not skip_band:
            sx = sig[key][1].values
            ax_xsec[r].fill_betweenx(ds_a.lat.values, xsec - Z_SIG*sx, xsec + Z_SIG*sx,
                                     color=colors[i], alpha=0.25, lw=0)
        ax_xsec[r].axvline(0, color='gray', ls='--')
        ax_xsec[r].set_xlim(-35, 35)
        ax_xsec[r].set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
        ax_xsec[r].grid(True, ls='--', alpha=0.5)
        ax_xsec[r].tick_params(axis='y', length=0)
        plt.setp(ax_xsec[r].get_yticklabels(), visible=False)

# --------------------------------------------------
plt.savefig('PLOTS/Figure7_refactored.png', dpi=500, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import cartopy.crs as ccrs

plt.rcParams['font.size'] = 12
plt.rcParams['hatch.color'] = 'k'
plt.rcParams['hatch.linewidth'] = 0.6

# ==================================================
# Settings for significance
# ==================================================
Z_SIG = 1.96   # 95% two-sided
N_EFF = 1      # keep 1 if *_std is already the uncertainty of the estimate
               # (e.g. from the block bootstrap); otherwise use number of years/members

# dataset ('A' or 'B') feeding each slot of compute_FC: (c, u_0, alpha, beta, qslwa)
SPEC = {
    'total': ('B', 'B', 'B', 'B', 'B'),
    'ab':    ('A', 'A', 'B', 'B', 'A'),   # alpha and beta together
    'u0':    ('A', 'B', 'A', 'A', 'A'),
    'cg':    ('B', 'A', 'A', 'A', 'A'),
    'q':     ('A', 'A', 'A', 'A', 'B'),
}
BASE = ('A',) * 5


# ==================================================
# Helpers
# ==================================================
def broadcast_zonal(field, ds_ref):
    return field.mean('lon').broadcast_like(ds_ref)


def _grad_coefs(dsets, spec, lat):
    """dF/d(elementary variable), keyed by (dataset, variable).
    Elementary variables: zgv, u_0, alpha, beta (c = zgv - u_0)."""
    sc, su, sa, sb, sq = spec
    c  = dsets[sc]['zgv'] - dsets[sc]['u_0']
    u0 = dsets[su]['u_0']
    al = dsets[sa]['alpha']
    be = dsets[sb]['beta']
    q  = dsets[sq]['qslwa']
    cosphi = np.cos(np.deg2rad(lat))

    u  = c + u0 + be - 2*al*q
    Fu = u*cosphi/(2*al)                              # dF/du
    Fa = -u**2*cosphi/(4*al**2) - 2*q*Fu              # dF/dalpha

    coef = {}
    def add(key, val):
        coef[key] = coef.get(key, 0) + val

    add((sc, 'zgv'),   Fu)    # through c = zgv - u0
    add((sc, 'u_0'),  -Fu)
    add((su, 'u_0'),   Fu)    # explicit u_0 slot
    add((sa, 'alpha'), Fa)
    add((sb, 'beta'),  Fu)
    return coef


def _lwa_mean(dd):
    """L_bar in beta = u_bar + alpha * L_bar. Use the stored field if present;
    otherwise recover it from beta_std = alpha_std * L_bar (signed!)."""
    if 'lwa_mean' in dd:
        return dd['lwa_mean']
    return dd['beta_std'] / dd['alpha_std']


def propagate_sigma(ds_a, ds_b, zonal):
    """
    First-order error propagation of the plotted differences.
    alpha and beta are fully correlated through beta = u_bar + alpha * L_bar, so
        d beta = L_bar * d alpha + d u_bar
    and the effective coefficient on the alpha error is  c_alpha + c_beta * L_bar.
    The independent part of beta (error on u_bar) is added separately if 'ubar_std'
    is available in the dataset.
    Returns {term: (sigma_map[lat,lon], sigma_xsec[lat])}.
    """
    ds   = {'A': ds_a, 'B': ds_b}
    lat  = ds_a['lat']
    nlon = ds_a.sizes['lon']
    ref  = ds_a['zgv']
    base = _grad_coefs(ds, BASE, lat)
    sq   = np.sqrt(N_EFF)

    out = {}
    for k, spec in SPEC.items():
        g = _grad_coefs(ds, spec, lat)
        var = 0
        for d in ('A', 'B'):
            dd = ds[d]

            def dc(v):                                  # coefficient difference for dataset d
                return g.get((d, v), 0) - base.get((d, v), 0)

            # alpha + the part of beta that follows alpha (fully correlated)
            c_eff = dc('alpha') + dc('beta') * _lwa_mean(dd)
            var = var + (c_eff * dd['alpha_std'] / sq)**2

            # independent part of beta (error on u_bar), if available
            if 'ubar_std' in dd:
                var = var + (dc('beta') * dd['ubar_std'] / sq)**2

            # remaining independent variables
            for v in ('zgv', 'u_0'):
                var = var + (dc(v) * dd[v + '_std'] / sq)**2

        var = var.broadcast_like(ref)

        # zonal mean of the difference, errors independent between longitudes
        sig_x = np.sqrt(var.mean('lon') / nlon)
        sig_m = sig_x.broadcast_like(ref) if zonal else np.sqrt(var)
        out[k] = (sig_m, sig_x)
    return out


def add_hatch(ax, lon, lat, diff, sigma, hatch='...', where='not_significant'):
    """Hatch significant points (default) or non-significant ones (where='not_significant')."""
    sig_mask = np.abs(diff) > Z_SIG*sigma
    if where == 'not_significant':
        # NaNs in diff/sigma are not hatched
        mask = (~sig_mask & diff.notnull() & sigma.notnull()).astype(float).values
    else:
        mask = sig_mask.astype(float).fillna(0).values
    if mask.any():
        ax.contourf(lon, lat, mask, levels=[0.5, 1.5], colors='none',
                    hatches=[hatch], transform=ccrs.PlateCarree())

# ==================================================
# Figure layout
# ==================================================
fig = plt.figure(figsize=(17, 17))

gs = gridspec.GridSpec(
    nrows=5,
    ncols=5,
    width_ratios=[6.0, 0.25, 1.2, 0.01, 6.0],
    hspace=0.45,
    wspace=-0.3
)

titles  = ['symmetric', 'asymmetric']
letters = ['(a)', '(b)', '(c)', '(d)', '(e)', '(f)', '(g)', '(h)', '(i)', '(l)']
colors  = ['k', 'silver']

coords = [(-75.0, 25.0), (-50.0, 25.0), (-50.0, 50.0)]
lons, lats = zip(*coords)
lons = np.array(lons)
lats = np.array(lats)

lons_closed = np.append(lons, lons[0])
lats_closed = np.append(lats, lats[0])

proj = ccrs.PlateCarree()

levels = np.arange(40, 400, 40)*3
levels_diff = np.arange(-42.75, 43, 1.5)

ax_xsec = []
term_keys = ['ab', 'u0', 'cg', 'q']

# ==================================================
# Loop symmetric / asymmetric
# ==================================================
for i in range(2):

    XSEC_COL = 2
    MAP_COL  = 0 if i == 0 else 4

    ds_a = ds_clims[i*2]
    ds_b = ds_clims[i*2+1]

    # error propagation for all terms of this case
    sig = propagate_sigma(ds_a, ds_b, zonal=(i == 0))

    # ==================================================
    # ROW 0 — TOTAL DIFFERENCE
    # ==================================================
    if i == 0:
        ax_xsec_total = fig.add_subplot(gs[0, XSEC_COL])

    ax_map = fig.add_subplot(gs[0, MAP_COL], projection=proj, sharey=ax_xsec_total)

    FC_b = compute_FC(ds_b['zgv']-ds_b['u_0'], ds_b['u_0'], ds_b['alpha'], ds_b['beta'], ds_b['qslwa'])
    FC_a = compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])

    if i == 0:
        FC_b = broadcast_zonal(FC_b, ds_a)
        FC_a = broadcast_zonal(FC_a, ds_a)

    total_diff = FC_b - FC_a

    ax_map.set_title(f'{letters[i]} Total difference – {titles[i]}')

    c1 = ax_map.contour(ds_a.lon, ds_a.lat, FC_a,
                        levels=levels, colors='dimgray')

    ax_map.contour(ds_b.lon, ds_b.lat, FC_b,
                   levels=levels, colors='dimgray',
                   linestyles='dashed')

    ax_map.clabel(c1, inline=True, colors='black', fontsize=11)

    sh = ax_map.contourf(ds_a.lon, ds_a.lat, total_diff,
                         levels=levels_diff,
                         cmap='RdBu_r', extend='both')

    # hatching: |dFc| > Z_SIG * sigma
    add_hatch(ax_map, ds_a.lon, ds_a.lat, total_diff, sig['total'][0])

    if i > 0:
        fig.colorbar(sh, ax=ax_map, label=r'$\Delta F_c$ [W m$^{-2}$]')

    if i > 0:
        ax_map.plot(lons_closed, lats_closed, '--', lw=3, color='red', transform=proj)

    ax_map.set_extent([ds_a.lon.min(), ds_a.lon.max(), 10, 85])
    ax_map.set_aspect(2.5)

    ax_map.set_xticks([-120, -60, 0, 60, 120], crs=proj)
    ax_map.set_xticklabels(['120°W', '60°W', '0°E', '60°E', '120°E'])
    ax_map.set_xlabel('Longitude')

    if i == 0:
        ax_map.set_yticks([20, 40, 60, 80])
        ax_map.set_yticklabels(['20°N', '40°N', '60°N', '80°N'])
        ax_map.set_ylabel('Latitude')
    else:
        plt.setp(ax_map.get_yticklabels(), visible=False)

    ax_map.gridlines(draw_labels=False, linestyle='--', alpha=0.6)

    xsec = total_diff.mean('lon').values

    ax_xsec_total.plot(xsec, ds_a.lat.values, color=colors[i], lw=2, label=titles[i])
    sx = sig['total'][1].values
    ax_xsec_total.fill_betweenx(ds_a.lat.values, xsec - Z_SIG*sx, xsec + Z_SIG*sx,
                                color=colors[i], alpha=0.25, lw=0)
    ax_xsec_total.axvline(0, color='gray', ls='--')
    ax_xsec_total.set_xlim(-35, 35)
    ax_xsec_total.set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
    ax_xsec_total.grid(True, ls='--', alpha=0.5)
    ax_xsec_total.tick_params(axis='y', length=0)
    plt.setp(ax_xsec_total.get_yticklabels(), visible=False)

    if i == 1:
        ax_xsec_total.legend(loc='upper right', bbox_to_anchor=(1.0, 1.35))

    # ==================================================
    # FC_terms definitions (order must match term_keys)
    # ==================================================
    FC_terms = [
        ('Alpha and beta contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_b['alpha'], ds_b['beta'], ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])),

        ('Reference wind contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_b['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])),

        ('Zonal group velocity contribution',
         compute_FC(ds_b['zgv']-ds_b['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa'])),

        ('Stationary LWA contribution',
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_b['qslwa']),
         compute_FC(ds_a['zgv']-ds_a['u_0'], ds_a['u_0'], ds_a['alpha'], ds_a['beta'], ds_a['qslwa']))
    ]

    # ==================================================
    # ROWS 1–4 — linear terms
    # ==================================================
    for r, (label, field_b, field_a) in enumerate(FC_terms):

        key = term_keys[r]

        if i == 0:
            ax_xsec.append(fig.add_subplot(gs[r+1, XSEC_COL]))

        ax_map = fig.add_subplot(gs[r+1, MAP_COL], projection=proj, sharey=ax_xsec[r])

        if i == 0:
            field_b = broadcast_zonal(field_b, ds_a)
            field_a = broadcast_zonal(field_a, ds_a)

        fc_diff = field_b - field_a

        ax_map.set_title(f'{letters[2*(r+1) + i]} {label} – {titles[i]}')

        c1 = ax_map.contour(ds_a.lon, ds_a.lat, field_a,
                            levels=levels, colors='dimgray')

        ax_map.contour(ds_b.lon, ds_b.lat, field_b,
                       levels=levels, colors='dimgray',
                       linestyles='dashed')

        ax_map.clabel(c1, inline=True, colors='black', fontsize=11)

        sh = ax_map.contourf(ds_a.lon, ds_a.lat, fc_diff,
                             levels=levels_diff,
                             cmap='RdBu_r', extend='both')

        # hatching: |dFc| > Z_SIG * sigma
        add_hatch(ax_map, ds_a.lon, ds_a.lat, fc_diff, sig[key][0])

        if i > 0:
            fig.colorbar(sh, ax=ax_map, label=r'$\Delta F_c$ [W m$^{-2}$]')

        if i > 0:
            ax_map.plot(lons_closed, lats_closed, '--', lw=3, color='red', transform=proj)

        ax_map.set_extent([ds_a.lon.min(), ds_a.lon.max(), 10, 85])
        ax_map.set_aspect(2.5)

        ax_map.set_xticks([-120, -60, 0, 60, 120], crs=proj)
        ax_map.set_xticklabels(['120°W', '60°W', '0°E', '60°E', '120°E'])
        ax_map.set_xlabel('Longitude')

        if i == 0:
            ax_map.set_yticks([20, 40, 60, 80])
            ax_map.set_yticklabels(['20°N', '40°N', '60°N', '80°N'])
            ax_map.set_ylabel('Latitude')
        else:
            plt.setp(ax_map.get_yticklabels(), visible=False)

        ax_map.gridlines(draw_labels=False, linestyle='--', alpha=0.6)

        xsec = fc_diff.mean('lon').values

        # stationary-LWA term is zero in the symmetric case by construction
        skip_band = (key == 'q' and i == 0)
        if skip_band:
            xsec = np.zeros_like(xsec)

        ax_xsec[r].plot(xsec, ds_a.lat.values, color=colors[i], lw=2, label=titles[i])
        if not skip_band:
            sx = sig[key][1].values
            ax_xsec[r].fill_betweenx(ds_a.lat.values, xsec - Z_SIG*sx, xsec + Z_SIG*sx,
                                     color=colors[i], alpha=0.25, lw=0)
        ax_xsec[r].axvline(0, color='gray', ls='--')
        ax_xsec[r].set_xlim(-35, 35)
        ax_xsec[r].set_xlabel(r'$\Delta F_c$ [W m$^{-2}$]')
        ax_xsec[r].grid(True, ls='--', alpha=0.5)
        ax_xsec[r].tick_params(axis='y', length=0)
        plt.setp(ax_xsec[r].get_yticklabels(), visible=False)

# --------------------------------------------------
plt.savefig('PLOTS/Figure7_refactored.png', dpi=500, bbox_inches='tight')
plt.show()
```
